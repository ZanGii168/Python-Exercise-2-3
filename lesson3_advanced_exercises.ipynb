## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Write your solution here
# Exercise 3: Operator precedence check
# Python precedence (highest to lowest): not > and > or

a = True
b = False
c = True
d = False

# 1) Complex expression WITHOUT extra parentheses
without_parens = not a and b or c and not d

# 2) Same expression WITH explicit parentheses
with_parens = ((not a) and b) or (c and (not d))

print("Without parentheses:", without_parens)
print("With parentheses:   ", with_parens)

# Comment: they MATCH, because the parentheses simply spell out the
# grouping Python already applies: not -> and -> or.
print("Match:", without_parens == with_parens)

# Bonus: parentheses that DISAGREE with the default precedence
x, y, z = False, True, True

default_result = x and y or z    # parsed as (x and y) or z -> True
forced_result = x and (y or z)   # different grouping       -> False

print()
print("x and y or z   :", default_result)
print("x and (y or z) :", forced_result)

# Comment: they do NOT match, so parentheses can change the result
# when they override the default precedence.
print("Match:", default_result == forced_result)
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here
# Two separate lists with the same contents
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True  -> same values
print(a is b)  # False -> two different objects in memory

# Assign one list to another variable (no copy is made)
c = a

print(a == c)  # True  -> same values
print(a is c)  # True  -> c is just another name for the same object

```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
# Write your solution here
items = ["python", "java", "go", "rust"]
queries = ["python", "go", "c++", "swift"]

# Using `in`
print("Using 'in':")
for q in queries:
    if q in items:
        print(f"{q}: found")
    else:
        print(f"{q}: not found")

# Using `not in`
print("\nUsing 'not in':")
for q in queries:
    if q not in items:
        print(f"{q}: not found")
    else:
        print(f"{q}: found")
```
